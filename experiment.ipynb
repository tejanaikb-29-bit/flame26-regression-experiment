# ScoreSense — FLAME'26 Regression Experiment

**Aim:** Demonstrate a regression workflow using a synthetic student-performance dataset. The dataset is generated for practice and is not real student data.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

DATA_PATH = 'data/student_performance.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

## 1. Explore the data
Check the dimensions, types, summary statistics, and missing values.

```python
print('Rows and columns:', df.shape)
display(df.dtypes.to_frame('data_type'))
display(df.describe().round(2))
print('\nMissing values per column:')
print(df.isna().sum())
```

## 2. Visualize a relationship
This scatter plot gives a first look at study time and the target score.

```python
plt.figure(figsize=(7, 4))
plt.scatter(df['study_hours_per_day'], df['final_exam_score'], alpha=0.7)
plt.xlabel('Study hours per day')
plt.ylabel('Final exam score')
plt.title('Study time vs. generated exam score')
plt.grid(alpha=0.2)
plt.show()
```

## 3. Split and train
We predict the continuous target `final_exam_score` from four input features.

```python
features = ['study_hours_per_day', 'attendance_percent', 'previous_exam_score', 'sleep_hours_per_day']
target = 'final_exam_score'
X = df[features]
y = df[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=26)
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print('Training rows:', len(X_train), '| Test rows:', len(X_test))
print('Intercept:', round(model.intercept_, 3))
print('Coefficients:')
for name, coefficient in zip(features, model.coef_):
    print(f'  {name}: {coefficient:.3f}')
```

## 4. Evaluate the model
MAE and RMSE measure typical prediction error; R² describes how much target variation is explained on the held-out test set.

```python
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)
print(f'MAE:  {mae:.2f} points')
print(f'RMSE: {rmse:.2f} points')
print(f'R²:   {r2:.3f}')
```

## 5. Actual vs. predicted

```python
plt.figure(figsize=(6, 6))
plt.scatter(y_test, predictions, alpha=0.75)
plt.plot([0, 100], [0, 100], linestyle='--')
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.xlabel('Actual score')
plt.ylabel('Predicted score')
plt.title('Actual vs. predicted scores')
plt.grid(alpha=0.2)
plt.show()
```

## 6. Try one sample
Change the values and run the cell to test a prediction. This is a learning demo, not an actual grade predictor.

```python
sample = pd.DataFrame([{
    'study_hours_per_day': 3.0,
    'attendance_percent': 85.0,
    'previous_exam_score': 70.0,
    'sleep_hours_per_day': 7.5,
}], columns=features)
print('Estimated score:', round(float(np.clip(model.predict(sample)[0], 0, 100)), 1), '/ 100')
```

## Conclusion
A Multiple Linear Regression model was trained on synthetic data, evaluated with MAE, RMSE and R², and used to produce a sample prediction. Since the dataset is artificial, results demonstrate the process rather than real-world academic performance.
